```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

x = np.array([-20, -10, -5, 0, 5, 10, 20, 25, 30], dtype=np.float32)
y = np.array([0, 0, 0, 0, 0, 0, 1, 1, 1], dtype=np.float32)

model = tf.keras.Sequential([
        tf.keras.layers.Dense(1, input_shape=(1,), activation='sigmoid') # Use input_shape instead of input_dim for clarity
        ])

model.compile(optimizer='sgd', loss='binary_crossentropy',
                               metrics=['binary_accuracy'])

model.fit(x, y, batch_size=1, epochs=300, shuffle=False)

# Reshape x explicitly for prediction to ensure it's 2D (num_samples, 1)
plt.plot(x, model.predict(x.reshape(-1, 1)), 'b', x, y, 'k.')

# Ensure the input array is float32 and reshaped for prediction
print(model.predict(np.array([1, 23, 50], dtype=np.float32).reshape(-1, 1)))
```
