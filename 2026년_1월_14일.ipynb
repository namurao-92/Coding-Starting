```python
from scipy import stats

x = [2, 3, 4, 5, 9, 10]
y = [30, 50, 60, 65, 80, 90]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def score_predict(x) :
    return slope * x + intercept

learning_time = int(input('예측하고 싶은 공부시간을 입력하세요: '))
score = score_predict(learning_time)

print(score)
```

```python
from scipy import stats

x = [1, 5, 10, 20]
y = [17, 25, 32, 40]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def time_predict(x) :
  return slope * x + intercept

distance = int(input('예측하고 싶은 배달시간에 거리를 입력하세요: '))
delivery_time = time_predict(distance)

print(delivery_time)
```
