```python
menu = ('새우버거', '치킨버거', '치즈버거')
price = (5000, 4000, 3000)

def showmenu() :
    print()
    print('[버거메뉴]')
    for i in range (0, len(menu)) :
        print(i+1, '.', menu[i], '\t가격 :', price[i])
    print()

def order(num) :
    burger = menu[num-1]
    print('주문버거 :', burger)

def pay(num, qty) :
    total_price = price[num-1] * qty
    return total_price

showmenu()

burger_num = int(input('주문할 버거 번호를 선택하세요: '))
order(burger_num)

burger_qty = int(input('\n버거 개수를 입력하세요 :'))
burger_total_price = pay(burger_num, burger_qty)

print('수량 : {}, 총금액 : {}'.format(burger_qty, burger_total_price))
ㅣㄴ
```

```python
score_data = {'김영희' : 70, '이영수' : 60, '최지성' : 90}

def print_name(score_data):
    for keys in score_data.keys() :
        if score_data[keys] > 70 : # Compare the score (value) instead of the name (key)
            print(keys)

print_name(score_data) # Call the function directly to avoid printing 'None'
```
