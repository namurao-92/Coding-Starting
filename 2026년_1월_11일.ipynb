```python
import random
import time

play = ['007게임','369게임','눈치게임','마피아게임']
print('게임 선택을 시작합니다. 1초 후에 알려줄께요')
time.sleep(1)
print('바로바로', random.choice(play))
```

```python
def count369(n) :
    count = 0
    q = n

    while q > 0 :
        r = q % 10
        q = q // 10
        if (r == 3) or (r == 6) or (r == 9) :
            count += 1

    return count

number = int(input('369게임! 숫자를 입력하세요 :'))

for i in range(1, number + 1) :
    n = count369(i)
    if n == 0 :
        print(i, end = ' ')
    else :
        print('X'*n, end=' ')
```
