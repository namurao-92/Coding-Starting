```python
postcode = input('우편번호 5자리 (04로 시작) :')

if not ( postcode[0] == '0' and postcode[1] == '4' ) :
    print("'04'로 시작하는 우편번호를 입력해 주세요.")

else :
    if postcode[2] in {'0', '1', '2'} :
        print ('마포구')
    elif postcode[2] in {'3', '4'} :
        print ('용산구')
    elif postcode[2] in {'5', '6'} :
        print ('중구')
    elif postcode[2] in {'7', '8'} :
        print ('성동구')
    elif postcode[2] in {'9'} :
        print ('광진구')
```

```python
import random

eng_dict = { '사과' : 'apple', '바나나' : 'banana', '복숭아' : 'peach',
             '오렌지' : 'orange', '딸기' : 'strawberry', '멜론' : 'melon',
             '학교' : 'school', '책' : 'book', '봄' : 'spring',
             '여름' : 'summer', '가을' : 'fall', '겨울' : 'winter'}

print('영어 단어 테스트 - 한글 뜻에 해당하는 영단어를 입력하세요')

while eng_dict :
    word = random.choice( list(eng_dict.keys()))

    while True :
        word_input = input('{} : '.format(word))
        if word_input == eng_dict[word] :
            eng_dict.pop(word)
            break
        else :
            print('!!! 다시 입력하세요.')

print('모든 단어의 테스트가 완료되었습니다')

```
