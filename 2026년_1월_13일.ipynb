```python
!pip install colab_turtle

import turtle as t
import random
import colab_turtle

# Initialize colab_turtle
colab_turtle.initializeTurtle()

def draw_star(x, y, color, size) :
    t.setheading(252)
    t.color(color)

    t.penup()
    t.goto(x, y)
    t.pendown()

    t.begin_fill()
    for i in range(5) :
        t.forward(size)
        t.left(144)

    t.end_fill()

n = int(input('별의 개수를 입력하세요 :'))

t.bgcolor('Black')
color_list = ['Red','Orange','Yellow','Green','Blue','Purple']

for i in range(n) :
    x = random.randint(-400, 400)
    y = random.randint(-300, 300)
    size = random.randint(30, 150)
    color = random.choice(color_list)

    draw_star(x, y, color, size)
```

```python
def binary_digit(num, digit) :
    q = num
    for i in range(digit + 1) :
        q = q//2

    if q == 1 :
        return True
    else :
        return False

power = int(input('질문 횟수를 입력하세요. (자연수) :'))
number = 0

for d in range(power) :

    for n in range(2*power) :
        if binary_digit(n, d) :
            print(n, end = ' ')

    ans = input(' - 생각한 숫자가 있나요? (Y/N) :')

    if ans == 'Y' or ans == 'y' :
        number += 2**d

print('number :', number)
```
